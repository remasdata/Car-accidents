## **1. Accuracy Dimension**


**Based on the data and its representation, an abnormal distribution appears. Therefore, the IQ test was chosen because it is not affected by outliers, as it depends on the median, not the mean or standard deviation.**

```python
import pandas as pd
import numpy as np

# Function to check outliers using IQR
def check_outliers_for_feature(df, feature):
    if feature not in df.columns:
        print(f"⚠️ Warning: Column '{feature}' not found in the dataset.")
        return

    if not np.issubdtype(df[feature].dtype, np.number):
        print(f"⚠️ Warning: Column '{feature}' is not numeric.")
        return

    # Compute statistical outliers using IQR
    Q1 = df[feature].quantile(0.25)
    Q3 = df[feature].quantile(0.75)
    IQR = Q3 - Q1
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    # Identify outliers
    iqr_outliers = df[(df[feature] < lower_bound) | (df[feature] > upper_bound)]

    # Print outlier
    print(f"🔍 Checking Accuracy (Outliers) for: {feature}")
    print(f"📌 IQR Outliers Found: {len(iqr_outliers)}")

    # Display sample outliers
    if not iqr_outliers.empty:
        print(f"\n🚩 IQR Outliers in {feature}:")
        print(iqr_outliers[[feature]].head(3))

# Check for outliers in the relevant features using IQR
check_outliers_for_feature(df, "Temperature(F)")
check_outliers_for_feature(df, "Humidity(%)")
check_outliers_for_feature(df, "Visibility(mi)")
check_outliers_for_feature(df, "Wind_Speed(mph)")

```

```python
from datetime import datetime

# Function to check date and time format
def check_datetime_format(value):
    try:

        datetime.strptime(str(value), '%Y-%m-%d %H:%M:%S')
        return True
    except:
        return False

# Apply validation to columns
df['Start_Valid'] = df['Start_Time'].apply(check_datetime_format)
df['End_Valid'] = df['End_Time'].apply(check_datetime_format)

# Extract rows containing invalid value
invalid_rows = df[~df['Start_Valid'] | ~df['End_Valid']]

print("Number of rows that do not fit the pattern:", len(invalid_rows))
df['Valid_Time_Range'] = df['Start_Time'] <= df['End_Time']
invalid_time_range = df[~df['Valid_Time_Range']]
print("Number of rows with invalid time range:", len(invalid_time_range))

```

```python
# Define a mapping for Severity == 4 (high severity) with corresponding weather conditions
def classify_weather_for_high_severity(row):
    if row['Severity'] == 4:
        # Assign a more severe weather condition based on logic
        if row['Weather_Condition'] in ['Clear', 'Fair']:
            return 'Thunderstorm'  # You can choose an appropriate weather type
        elif row['Weather_Condition'] in ['Cloudy', 'Mostly Cloudy', 'Partly Cloudy']:
            return 'Rain'
        elif row['Weather_Condition'] == 'Overcast':
            return 'Snow'
        elif row['Weather_Condition'] in ['Light Rain', 'Drizzle']:
            return 'Thunderstorm'
        elif row['Weather_Condition'] in ['Fog', 'Haze']:
            return 'Mist'
        elif row['Weather_Condition'] == 'Scattered Clouds':
            return 'Thunder'
        else:
            return row['Weather_Condition']  # Keep original weather if it's already severe
    else:
        return row['Weather_Condition']  # If not severe, keep original weather

# Apply the classification function to the dataframe
df['Weather_Condition'] = df.apply(classify_weather_for_high_severity, axis=1)

# Check the results
inconsistent_severity_weather = df[(df['Severity'] == 4) & (df['Weather_Condition'].isin(['Clear', 'Fair']))]
print("❌ High Severity cases with clear weather after correction:", len(inconsistent_severity_weather))
```

```python
#Fill-in Missing Values in Categorical Columns
df = df.dropna(subset=["City"])
# Create a mapping from City → most frequent Weather_Condition
city_weather_map = (
    df.groupby("City")["Weather_Condition"]                             # Group the data by city, selecting the 'Weather_Condition' column
    .agg(lambda x: x.mode()[0] if not x.mode().empty else np.nan)       # Aggregate within each group
    .to_dict()                                                          # Use the mode (most frequent value) as the representative Weather_Condition
)                                                                       # If no mode (empty group), return NaN to avoid errors
 # Fill missing 'climate' values using the map                                                                       # Convert the result into a dictionary: {city: Weather_Condition}
df["Weather_Condition"] = df["City"].map(city_weather_map)

df["Sunrise_Sunset"] = df["Sunrise_Sunset"].fillna(df["Sunrise_Sunset"].mode()[0])
```

.The rows containing missing values in the City column were removed to ensure the accuracy of linking the city with the weather condition.

.The missing values in Weather_Condition were filled based on the most common weather condition (mode) in each city

```python
# medical_supply (symmetric): Predictive Imputation (KNN):
from sklearn.impute import KNNImputer
# columns that contain missing data
cols_to_impute = ['Start_Time', 'End_Time']
#Filling in the missing values using KNN
imputer = KNNImputer(n_neighbors = 3)
df[cols_to_impute] = imputer.fit_transform(df[cols_to_impute])
#Calculate the duration by subtracting the start time from the end time.
df["Duration"] = df["End_Time"] - df["Start_Time"]

#Impute any missing values using the median.
df["Duration"] = df["Duration"].fillna(df["Duration"].median())


```

```python
# Step 5: pattern mining algorithm  Apriori - FP-Growth


# Import algorithms to find frequent itemsets
from mlxtend.frequent_patterns import apriori, fpgrowth

# Set the minimum support threshold (e.g., 0.7 )
min_support = 0.7

#  Apply Apriori algorithm
frequent_itemsets_apriori = apriori(
    df_encoded,               # use one-hot encoded DataFrame
    min_support=min_support,  # apply minimum support threshold
    use_colnames=True         # ensures that the output uses actual item names instead of column indices
)

#  Apply FP-Growth algorithm
frequent_itemsets_fpgrowth = fpgrowth(
    df_encoded,
    min_support=min_support,
    use_colnames=True
)
frequent_itemsets_fpgrowth.head()
frequent_itemsets_apriori.head()
#len(frequent_itemsets_fpgrowth)
```

Using two algorithms to analyze recurring patterns within the data, which are:
Apriori Algorithm

FP-Growth Algorithm

A minimum support threshold of 0.7 was used, meaning we are only looking for patterns that appeared in 70% or more of the data.

```python
# Step 6: Filter and inspect frequent itemsets.

# Filter 2-itemsets from Apriori output
apriori_2_itemsets = frequent_itemsets_apriori[
    frequent_itemsets_apriori['itemsets'].apply(lambda x: len(x) == 2) # keep only itemsets that contain at least 2 items
].sort_values(by='support', ascending=False)

# Filter 2-itemsets from FP-Growth output
fpgrowth_2_itemsets = frequent_itemsets_fpgrowth[
    frequent_itemsets_fpgrowth['itemsets'].apply(lambda x: len(x) == 2)
].sort_values(by='support', ascending=False)

# Temporarily allow wider column display
pd.set_option('display.max_colwidth', None)

# Display top 5 from each
print(f"Total 2-itemsets from Apriori: {len(apriori_2_itemsets)}")
print("Top five 2-itemsets:")
display(apriori_2_itemsets.head(5))

print(f"\n\nTotal 2-itemsets from FP-Growth: {len(fpgrowth_2_itemsets)}")
print("Top five 2-itemsets:")
display(fpgrowth_2_itemsets.head(5))
```

```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier, plot_tree
from sklearn.preprocessing import LabelEncoder
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix, ConfusionMatrixDisplay


target_col = 'Severity'

features = [
    'Temperature(F)', 'Humidity(%)', 'Visibility(mi)', 'Wind_Speed(mph)',
    'Weather_Condition', 'Turning_Loop', 'Junction', 'No_Exit',
    'Traffic_Signal', 'Sunrise_Sunset', 'City', 'State'
]

X = df[features].copy()
y = df[target_col]



binary_cols = ['Turning_Loop', 'Junction', 'No_Exit', 'Traffic_Signal']
for col in binary_cols:
    X[col] = X[col].map({'Yes': 1, 'No': 0})

le_weather = LabelEncoder()
X['Weather_Condition'] = le_weather.fit_transform(X['Weather_Condition'])

le_sun = LabelEncoder()
X['Sunrise_Sunset'] = le_sun.fit_transform(X['Sunrise_Sunset'])

le_city = LabelEncoder()
X['City'] = le_city.fit_transform(X['City'])

le_state = LabelEncoder()
X['State'] = le_state.fit_transform(X['State'])


X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, random_state=1, stratify=y
)

tree = DecisionTreeClassifier(max_depth=6, criterion='entropy', random_state=1)
tree.fit(X_train, y_train)
y_pred_tree = tree.predict(X_test)


print("\n- Decision Tree Accuracy:", round(accuracy_score(y_test, y_pred_tree)*100, 2), "%")
print("\n- Classification Report:\n", classification_report(y_test, y_pred_tree))
print("\n- Confusion Matrix:")
cm = confusion_matrix(y_test, y_pred_tree)
ConfusionMatrixDisplay(confusion_matrix=cm).plot()
plt.show()
```
